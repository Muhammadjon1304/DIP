## Setup: Import Libraries

```python
import cv2
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
print('✓ Libraries imported')
```

## Load and Analyze Image

```python
image_path = '/Users/muhammadjonparpiyev/Documents/DIP/Week 4/Lab 1/anfield.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

print('IMAGE ANALYSIS: anfield.jpg')
print(f'Shape: {image.shape}')
print(f'Min: {image.min()}, Max: {image.max()}')
print(f'Mean: {image.mean():.2f}, Std: {image.std():.2f}')
```

## Task 1: Power Law Transformation (Gamma Correction)

```python
def gamma_correction(image, c, gamma):
    I_norm = image.astype('float32') / 255
    I_transformed = c * (I_norm ** gamma)
    return np.clip(I_transformed, 0, 1)

gamma_values = [0.33, 0.5, 0.67, 1.0, 1.5, 2.0, 3.0]
c = 1.0

fig, axes = plt.subplots(2, 4, figsize=(18, 10))
axes = axes.flatten()

axes[0].imshow(image, cmap='gray')
axes[0].set_title('Original', fontsize=12, fontweight='bold')
axes[0].axis('off')

for idx, gamma in enumerate(gamma_values, 1):
    corrected = gamma_correction(image, c, gamma)
    corrected_8bit = (corrected * 255).astype(np.uint8)
    axes[idx].imshow(corrected_8bit, cmap='gray')
    axes[idx].set_title(f'γ={gamma}', fontsize=11, fontweight='bold')
    axes[idx].axis('off')

plt.tight_layout()
plt.savefig('/Users/muhammadjonparpiyev/Documents/DIP/Week 4/anfield_gamma_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('✓ Gamma comparison saved')
```

## Apply Optimal Gamma

```python
best_gamma = 0.67
corrected_image = gamma_correction(image, 1.0, best_gamma)
corrected_8bit = (corrected_image * 255).astype(np.uint8)

print(f'Optimal γ = {best_gamma}')
print(f'Original mean: {image.mean():.2f} → Corrected mean: {corrected_8bit.mean():.2f}')

fig2, axes2 = plt.subplots(1, 2, figsize=(14, 6))
axes2[0].imshow(image, cmap='gray')
axes2[0].set_title('Original', fontsize=13, fontweight='bold')
axes2[0].axis('off')

axes2[1].imshow(corrected_8bit, cmap='gray')
axes2[1].set_title(f'Gamma Corrected (γ={best_gamma})', fontsize=13, fontweight='bold')
axes2[1].axis('off')

plt.tight_layout()
plt.savefig('/Users/muhammadjonparpiyev/Documents/DIP/Week 4/anfield_gamma_result.png', dpi=150, bbox_inches='tight')
plt.show()

cv2.imwrite('/Users/muhammadjonparpiyev/Documents/DIP/Week 4/anfield_gamma_corrected.jpg', corrected_8bit)
print('✓ Gamma result saved')
```

## Task 2: Contrast Stretching

```python
def contrast_stretching(image, smax, smin):
    rmin, rmax = image.min(), image.max()
    if rmax == rmin:
        return np.ones_like(image) * smin
    I_stretched = ((smax - smin) / (rmax - rmin)) * (image - rmin) + smin
    return np.clip(I_stretched, smin, smax)

param_combinations = [(255, 0), (255, 50), (255, 100), (160, 100), (200, 50)]

fig3, axes3 = plt.subplots(2, 3, figsize=(16, 10))
axes3 = axes3.flatten()

axes3[0].imshow(image, cmap='gray')
axes3[0].set_title('Original', fontsize=12, fontweight='bold')
axes3[0].axis('off')

for idx, (smax, smin) in enumerate(param_combinations, 1):
    stretched = contrast_stretching(image.astype('float32'), smax, smin)
    stretched_8bit = stretched.astype(np.uint8)
    axes3[idx].imshow(stretched_8bit, cmap='gray')
    axes3[idx].set_title(f'smax={smax}, smin={smin}', fontsize=11, fontweight='bold')
    axes3[idx].axis('off')

plt.tight_layout()
plt.savefig('/Users/muhammadjonparpiyev/Documents/DIP/Week 4/anfield_contrast_stretching_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
print('✓ Contrast stretching comparison saved')
```

## Normalized vs Non-normalized Comparison

```python
stretched_255_0 = contrast_stretching(image.astype('float32'), 255, 0).astype(np.uint8)
normalized_image = image.astype('float32') / 255
stretched_norm = contrast_stretching(normalized_image, 1.0, 0.0)
stretched_norm_8bit = (stretched_norm * 255).astype(np.uint8)

print('Normalized vs Non-normalized:')
print(f'  Max difference: {np.max(np.abs(stretched_255_0.astype(float) - stretched_norm_8bit.astype(float))):.2f}')
print(f'  Result: IDENTICAL outputs')

fig4, axes4 = plt.subplots(1, 3, figsize=(15, 5))
axes4[0].imshow(image, cmap='gray')
axes4[0].set_title('Original', fontsize=12, fontweight='bold')
axes4[0].axis('off')

axes4[1].imshow(stretched_255_0, cmap='gray')
axes4[1].set_title('Non-normalized\n(smax=255, smin=0)', fontsize=12, fontweight='bold')
axes4[1].axis('off')

axes4[2].imshow(stretched_norm_8bit, cmap='gray')
axes4[2].set_title('Normalized\n(smax=1, smin=0)', fontsize=12, fontweight='bold')
axes4[2].axis('off')

plt.tight_layout()
plt.savefig('/Users/muhammadjonparpiyev/Documents/DIP/Week 4/anfield_normalized_vs_nonnormalized.png', dpi=150, bbox_inches='tight')
plt.show()
print('✓ Comparison visualization saved')
```
