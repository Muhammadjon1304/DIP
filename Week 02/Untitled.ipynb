```python
import cv2
import matplotlib.pyplot as plt
# Read a grayscale image
image = cv2.imread('/Users/muhammadjonparpiyev/Documents/licensed-image.jpeg')
image_RGB = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
# Subtract an intensity value from all pixels
subtracted_image = cv2.subtract(image_RGB, 100)
# Display original and subtracted images
plt.subplot(1, 2, 1)
plt.imshow(image_RGB)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(subtracted_image)
plt.title('Subtracted Image (-100)')
```
