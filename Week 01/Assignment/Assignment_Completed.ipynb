## 🔹 Part 1: Using PIL (Pillow)

PIL loads images as **PIL Image objects**, not NumPy arrays.

```python
# TODO 1:
# 1. Open the image "images/test_image.jpg" using PIL
# 2. Print the image type, size, and mode
# 3. Display the image

from PIL import Image

# Open the image using PIL
img_pil = Image.open('images.jpeg')

# Print image properties
print(f"Image type: {type(img_pil)}")
print(f"Image size: {img_pil.size}")
print(f"Image mode: {img_pil.mode}")

# Display the image
img_pil.show()
```

### ✨ Task 1.1: PIL → NumPy Conversion

```python
# TODO 2:
# Convert the PIL image to a NumPy array
# Print the array shape and data type

import numpy as np

# Convert PIL image to NumPy array
img_array = np.array(img_pil)

# Print array properties
print(f"Array shape: {img_array.shape}")
print(f"Array data type: {img_array.dtype}")
```

## 🔹 Part 2: Matplotlib Image Reading

Matplotlib reads images directly as NumPy arrays.

```python
# TODO 3:
# 1. Read the image using matplotlib.image.imread
# 2. Print its shape
# 3. Display it using plt.imshow

import matplotlib.image as mpimg
import matplotlib.pyplot as plt

# Read image using matplotlib
img_mpl = mpimg.imread('images.jpeg')

# Print shape
print(f"Image shape: {img_mpl.shape}")

# Display the image
plt.figure(figsize=(8, 6))
plt.imshow(img_mpl)
plt.axis('off')
plt.title("Image loaded with Matplotlib")
plt.show()
```

### ❓ Concept Question
**Q:** Why does Matplotlib treat images as NumPy arrays?

**Answer:** Matplotlib treats images as NumPy arrays because this format allows for efficient numerical operations, compatibility with scientific computing tools, and flexible manipulation of pixel data. NumPy arrays enable vectorized operations that are much faster than iterating through individual pixels, making them ideal for image processing tasks.

## 🔹 Part 3: scikit-image

scikit-image works with well-defined data ranges.

```python
# TODO 4:
# 1. Read the image using skimage.io.imread
# 2. Convert it to float using img_as_float
# 3. Print min and max pixel values
# 4. Display the image

from skimage import io, img_as_float
import matplotlib.pyplot as plt

# Read image using scikit-image
img_ski = io.imread('images.jpeg')

# Convert to float
img_float = img_as_float(img_ski)

# Print min and max pixel values
print(f"Min pixel value: {img_float.min()}")
print(f"Max pixel value: {img_float.max()}")

# Display the image
plt.figure(figsize=(8, 6))
plt.imshow(img_float)
plt.axis('off')
plt.title("Image loaded with scikit-image (float)")
plt.show()
```

### ✨ Task 3.1: Data Type Exploration

**Q:** Why is using `astype(float)` discouraged compared to `img_as_float`?

**Answer:** `img_as_float` is preferred because it properly normalizes pixel values to the [0, 1] range, handles different input data types correctly, and is specifically designed for image processing. Using `astype(float)` just converts data type without proper scaling (e.g., 255 stays as 255 instead of becoming 1.0), which can cause incorrect calculations and visualization issues.

## 🔹 Part 4: OpenCV

⚠️ OpenCV loads color images in **BGR** format.

```python
# TODO 5:
# 1. Read the image in grayscale and color using OpenCV
# 2. Convert the color image from BGR to RGB
# 3. Display the RGB image using matplotlib

import cv2
import matplotlib.pyplot as plt

# Read image in grayscale
img_gray = cv2.imread('images.jpeg', cv2.IMREAD_GRAYSCALE)

# Read image in color (default BGR)
img_bgr = cv2.imread('images.jpeg')

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Display the RGB image
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.title("Grayscale Image")

plt.subplot(1, 2, 2)
plt.imshow(img_rgb)
plt.axis('off')
plt.title("RGB Image")
plt.show()
```

### ✨ Task 4.1: Image Processing

Apply **Canny edge detection** on the grayscale image and display the result.

```python
# TODO 6:
# Perform edge detection using cv2.Canny

# Apply Canny edge detection
edges = cv2.Canny(img_gray, 100, 200)

# Display edge detection result
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.title("Original Grayscale")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.title("Canny Edge Detection")
plt.show()
```
